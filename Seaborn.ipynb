```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.get_dataset_names()
```

```python
penguins=sns.load_dataset("penguins")
penguins.head()
```

```python
penguins["species"].value_counts()
```

```python
penguins["island"].value_counts()
```

```python
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island")
```

```python
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="sex")
```

```python
#The types of styling themes are as follows : white, whitegrid, dark, darkgrid, ticks

#set_context allows us to change default size of graphs parameters like index, labels, etc.
#Types are paper-Smallest, notebook-Medium, talk-Large, poster-Very Large

#There are default palletes in seaborn
#Types are:
#Set1-Bright,High contrast
#Set2-Soft,Pastel like colors
#Set3-A mix of soft and bold varied hues
#It also has parameters like deep muted bright pastel dark colorblind
```

```python
#Scatter plot
sns.set_style("ticks")
sns.set_context("notebook")
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island",palette="Set1")
sns.despine()
```

```python
sns.scatterplot(data=penguins,x="species",y="body_mass_g",hue="island",style="sex")
```

```python
#Strip plot
sns.stripplot(data=penguins,x="species",y="body_mass_g",hue="island",dodge=True,jitter=True)
```

```python
#Swarm Plot
sns.swarmplot(data=penguins,x="species",y="body_mass_g",hue="island")
```

```python
sns.histplot(data=penguins,x="body_mass_g",hue="sex",multiple="stack")
sns.set_style("whitegrid")
sns.despine()
```

```python
#Regression Plot
sns.regplot(data=penguins,y="flipper_length_mm",x="body_mass_g",scatter=False)
```

```python
#Line Plot
sns.lineplot(data=penguins,x="body_mass_g",y="flipper_length_mm",hue="island",style="sex")
sns.set_style("whitegrid")
```

```python
#Joint Plot
sns.jointplot(data=penguins,x="body_mass_g",y="flipper_length_mm")
```

```python
#Joint Plot-2
sns.jointplot(data=penguins,x="body_mass_g",y="flipper_length_mm",hue="island",kind="kde",palette="Set1")
```

```python
#Bar Plot
sns.barplot(data=penguins,x="species",y="body_mass_g",hue="sex",palette="Set1",estimator=np.sum)#estimators could also be mean vr std
```

```python
#Just a tp plot but must see
count = penguins["species"].value_counts()
sns.barplot(x=count.index, y=count.values)
plt.xlabel("Species")
plt.ylabel("Number of Penguins")
plt.title("Number of Penguins by Species")
plt.show()
```

```python
#Count Plot
sns.countplot(data=penguins,x="species",hue="island",palette="Set2")
```

```python
#Box Plot
sns.boxplot(data=penguins,x="species",y="body_mass_g",hue="sex")
```

```python
#Violin Plot
sns.violinplot(data=penguins,x="species",y="body_mass_g",hue="sex",split=True)
```

```python
sns.violinplot(data=penguins, x="species", y="body_mass_g", palette="pastel")
sns.swarmplot(data=penguins, x="species", y="body_mass_g", hue="island", size=3, palette="dark")
```

```python
# KDE -> for estimating the probability density function of continuous or non-parametric data
# Or in simple terms:
# a smooth curve that shows the distribution of data
# It helps you see where the data is concentrated (peaks) and how it spreads.

sns.kdeplot(data=penguins,x="body_mass_g",hue="species",fill=True)
```

```python
# Heatmap: a 2D color-coded matrix that shows relationships or patterns between two variables.
# It's commonly used to:
# - Show correlation between features
# - Visualize confusion matrices -> will see in the later stages.
# - Highlight missing data -> sns.heatmap(df.isnull())
columns=["bill_length_mm","bill_depth_mm","flipper_length_mm","body_mass_g"]
sns.heatmap(data=penguins[columns].corr(),cmap="Blues",annot=True,linewidth=2,linecolor="black",vmin=-0.2)
```

```python
#Rug Plot
sns.rugplot(data=penguins,x="body_mass_g",hue="species",height=1,palette="Set1")
```

```python
sns.pairplot(data=penguins,hue="sex",diag_kind="kde",palette="Set1")
```

```python
#Pair Grid--> subplot grid for plotting partwise relationships in a dataset.
graph = sns.PairGrid(data=penguins, hue="sex", palette="Set2")
graph.map_upper(sns.scatterplot)  # upper triangle
graph.map_lower(sns.kdeplot)      # lower triangle
graph.map_diag(sns.histplot)      # diagonal
graph.add_legend()
```
